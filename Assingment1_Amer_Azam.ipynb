```python
# String.upper()
# Method .upper() menukar input negeri (e.g., "Johor", "johor", "JOHOR") menjadi HURUF BESAR sepenuhnya.

Tag_negeri = "negeri sembilan"
Tag_negeri2 = "melaka"
Tag_negeri3 = "johor"

print(f"Tag Negeri (Upper): {Tag_negeri.upper()}")
print(f"Tag Negeri2 (Upper):{Tag_negeri2.upper()}")
print(f"Tag Negeri3 (Upper): {Tag_negeri3.upper()}")
```

```python
# String .strip()
# Method .strip() membuang zero space like a gap left and right.
pdf_1 = "   Soalan_Trial_Set1.pdf   "
pdf_2 = " Soalan_Trial_Set2.pdf "
pdf_3 = "   Soalan_Trial_Set3.pdf"



print(f"'{pdf_1.strip()}'")
print(f"'{pdf_2.strip()}'")
print(f"'{pdf_3.strip()}'")
```

```python
import time

print("PROGRAM LATIHAN STRING PYTHON")
time.sleep(1)


ayat = input("\nMasukkan ayat atau nama fail untuk diedit: ")

pilihan = int(input("Sila pilih nombor method yang nak digunakan:1. HURUF BESAR (.upper)2. huruf kecil (.lower) 3. Huruf Tajuk (.title) 4. Buang Space Tepi (.strip) 5. Ganti Perkataan (.replace) Plase Only Usee (1-5):" ))

print("\nResult")



if pilihan == 1:

    hasil = ayat.upper()
    print(f"Hasil Huruf Besar: {hasil}")

elif pilihan == 2:
    hasil = ayat.lower()
    print(f"Hasil Huruf Kecil: {hasil}")

elif pilihan == 3:
    hasil = ayat.title()
    print(f"Hasil Huruf Tajuk: {hasil}")

elif pilihan == 4:
    hasil = ayat.strip()
    print(f"Hasil Tanpa Space Tepi: '{hasil}'")

elif pilihan == 5:
    asal = input("Perkataan mana nak tukar? ")
    baru = input("Tukar kepada apa? ")
    hasil = ayat.replace(asal, baru)
    print(f"Ayat Baru: {hasil}")

else: print("WRONG ,  1-5 only.")

print("END")
```

```python
import threading
from queue import Queue, Empty

def get_input(q):
    user_input = input("Masukkan nombor: ")
    q.put(user_input)

q = Queue()
thread = threading.Thread(target=get_input, args=(q,))
thread.daemon = True
thread.start()

try:
    input_data = q.get(timeout=7)
    print(f"Input received: {input_data}")
    nombor = int(input_data)
    if nombor % 2 == 0:
        print("This is an EVEN number.")
    else:
        print("This is an ODD number.")
except Empty:
    print("END (timeout)")
except ValueError:
    print("END (bukan nombor)")
```
